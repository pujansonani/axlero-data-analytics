```python
import pandas as pd

file_path = r"C:\Users\darsh\OneDrive\Axlero-Data-Analytics-Project\Data\raw\credit_risk_dataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
print("Dataset Columns:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_report = pd.DataFrame({
    'Missing Values': missing_count,
    'Missing Percentage': missing_percentage
})

print(missing_report)
```

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python
duplicate_records = df[df.duplicated(keep=False)]

print("Total duplicate records:", len(duplicate_records))

duplicate_records.head(10)
```

```python
df_clean = df.copy()

df_clean = df_clean.drop_duplicates()

print("Original Dataset Shape:", df.shape)
print("After Removing Duplicates:", df_clean.shape)
print("Remaining Duplicates:", df_clean.duplicated().sum())
```

```python
print("Age Statistics:")
print(df_clean['person_age'].describe())

print("\nEmployment Length Statistics:")
print(df_clean['person_emp_length'].describe())

print("\nLoan Amount Statistics:")
print(df_clean['loan_amnt'].describe())
```

```python
print("Records with Age Above 100:")
print(df_clean[df_clean['person_age'] > 100])

print("\nRecords with Employment Length Above 60:")
print(df_clean[df_clean['person_emp_length'] > 60])
```

```python
# Check records with suspicious age
age_issues = df_clean[df_clean['person_age'] > 100]

# Check records with suspicious employment length
employment_issues = df_clean[df_clean['person_emp_length'] > 60]

print("Suspicious Age Records:", len(age_issues))
print("Suspicious Employment Records:", len(employment_issues))

# Check whether these records overlap
print("\nOverlapping Records:")
print(
    df_clean[
        (df_clean['person_age'] > 100) |
        (df_clean['person_emp_length'] > 60)
    ]
)
```

```python
# Replace invalid age values with NaN
df_clean.loc[df_clean['person_age'] > 100, 'person_age'] = pd.NA

# Replace invalid employment length values with NaN
df_clean.loc[df_clean['person_emp_length'] > 60, 'person_emp_length'] = pd.NA

print("Invalid age values:", df_clean['person_age'].isna().sum())
print("Invalid employment values:", df_clean['person_emp_length'].isna().sum())
```

```python
# Check updated missing values
missing_values = df_clean.isnull().sum()

missing_summary = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': (missing_values / len(df_clean)) * 100
})

print(missing_summary)
```

```python
# Check missing interest rates by loan grade

print(
    df_clean.groupby('loan_grade')['loan_int_rate']
    .agg(['count', 'mean', 'median'])
)

print("\nMissing Interest Rates by Loan Grade:")

print(
    df_clean[df_clean['loan_int_rate'].isna()]
    ['loan_grade']
    .value_counts()
    .sort_index()
)
```

```python
# Fill missing interest rates using loan-grade-wise median

df_clean['loan_int_rate'] = (
    df_clean.groupby('loan_grade')['loan_int_rate']
    .transform(lambda x: x.fillna(x.median()))
)

print("Missing loan interest rates after imputation:",
      df_clean['loan_int_rate'].isna().sum())
```

```python
print(df_clean.groupby('loan_grade')['loan_int_rate'].median())
```

```python
# Check employment length statistics

print("Overall Median:",
      df_clean['person_emp_length'].median())

print("\nMedian Employment Length by Loan Grade:")

print(
    df_clean.groupby('loan_grade')['person_emp_length']
    .median()
)
```

```python
# Fill missing employment length using overall median

employment_median = df_clean['person_emp_length'].median()

df_clean['person_emp_length'] = (
    df_clean['person_emp_length'].fillna(employment_median)
)

print("Median used:", employment_median)

print("Missing employment values after imputation:",
      df_clean['person_emp_length'].isna().sum())
```

```python
print("Age Median:", df_clean['person_age'].median())

print("\nAge Statistics:")
print(df_clean['person_age'].describe())
```

```python
# Fill missing age values using median

age_median = df_clean['person_age'].median()

df_clean['person_age'] = (
    df_clean['person_age'].fillna(age_median)
)

print("Median age used:", age_median)

print("Missing age values after imputation:",
      df_clean['person_age'].isna().sum())
```

```python
print("Final Missing Values:")
print(df_clean.isnull().sum())

print("\nTotal Missing Values:",
      df_clean.isnull().sum().sum())
```

```python
# Check target column values

print("Unique Loan Status Values:")
print(df_clean['loan_status'].unique())

print("\nLoan Status Distribution:")
print(df_clean['loan_status'].value_counts())

print("\nMissing Target Values:",
      df_clean['loan_status'].isnull().sum())
```

```python
# Check unique values in categorical columns

categorical_columns = [
    'person_home_ownership',
    'loan_intent',
    'loan_grade',
    'cb_person_default_on_file'
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(df_clean[column].unique())
```

```python
# Check negative values in numeric columns

numeric_columns = [
    'person_age',
    'person_income',
    'person_emp_length',
    'loan_amnt',
    'loan_int_rate',
    'loan_percent_income',
    'cb_person_cred_hist_length'
]

for column in numeric_columns:
    print(f"\n{column}")
    print("Minimum:", df_clean[column].min())
    print("Maximum:", df_clean[column].max())
    print("Negative values:", (df_clean[column] < 0).sum())
```

```python
# Check employment length against person's age

employment_age_issues = df_clean[
    df_clean['person_emp_length'] > df_clean['person_age']
]

print("Employment Length Greater Than Age:")
print(len(employment_age_issues))

# Check credit history length against person's age

credit_age_issues = df_clean[
    df_clean['cb_person_cred_hist_length'] > df_clean['person_age']
]

print("\nCredit History Longer Than Age:")
print(len(credit_age_issues))

# Check zero income

print("\nZero Income Records:",
      (df_clean['person_income'] == 0).sum())
```

```python
# Calculate loan-to-income ratio

calculated_ratio = (
    df_clean['loan_amnt'] / df_clean['person_income']
)

# Compare calculated ratio with existing column

ratio_diff = (
    calculated_ratio - df_clean['loan_percent_income']
).abs()

print("Ratio Difference Summary:")
print(ratio_diff.describe())

print("\nDifferences Above 0.01:",
      (ratio_diff > 0.01).sum())
```

```python
# Identify records with significant ratio differences

ratio_issues = df_clean.loc[
    ratio_diff > 0.01,
    [
        'person_income',
        'loan_amnt',
        'loan_percent_income'
    ]
].copy()

ratio_issues['calculated_ratio'] = (
    ratio_issues['loan_amnt'] / ratio_issues['person_income']
)

ratio_issues['difference'] = (
    ratio_issues['calculated_ratio']
    - ratio_issues['loan_percent_income']
).abs()

print("Number of suspicious records:", len(ratio_issues))

display(ratio_issues.head(10))
```

```python
# Calculate and round loan-to-income ratio

expected_ratio = (
    df_clean['loan_amnt'] / df_clean['person_income']
).round(2)

# Compare with existing ratio column

rounded_mismatches = (
    expected_ratio != df_clean['loan_percent_income']
)

print("Mismatched records after rounding:",
      rounded_mismatches.sum())

print("Total records:", len(df_clean))
```

```python
# Get mismatched records

mismatch_data = df_clean.loc[
    rounded_mismatches,
    [
        'person_income',
        'loan_amnt',
        'loan_percent_income'
    ]
].copy()

mismatch_data['expected_ratio'] = (
    df_clean.loc[rounded_mismatches, 'loan_amnt']
    / df_clean.loc[rounded_mismatches, 'person_income']
).round(2)

print("First 15 mismatched records:")

display(mismatch_data.head(15))
```

```python
# Check calculated ratio range

calculated_ratio = (
    df_clean['loan_amnt'] / df_clean['person_income']
)

print("Minimum calculated ratio:", calculated_ratio.min())
print("Maximum calculated ratio:", calculated_ratio.max())

print("\nExisting ratio range:")
print(df_clean['loan_percent_income'].min())
print(df_clean['loan_percent_income'].max())
```

```python
# Final ratio validation summary

print("Total Records:", len(df_clean))

print("Ratio mismatches:", rounded_mismatches.sum())

print("Ratio matching records:",
      (~rounded_mismatches).sum())

print("Original ratio column preserved.")
```

```python
# Final dataset validation

print("Final Dataset Shape:", df_clean.shape)

print("Duplicate Rows:", df_clean.duplicated().sum())

print("\nTotal Missing Values:",
      df_clean.isnull().sum().sum())

print("\nMissing Values by Column:")
print(df_clean.isnull().sum())
```

```python
# Save final validated dataset

output_path = r"C:\Users\darsh\OneDrive\Axlero-Data-Analytics-Project\Data\processed\final_validated_dataset.csv"

df_clean.to_csv(output_path, index=False)

print("Final dataset saved successfully!")
print("Final shape:", df_clean.shape)
print("Saved location:", output_path)
```
